```python
# question3
arr=[21, 10, 45, 32, 67, 5, 18]
#Linear Search (find 32)
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i   # index where found
    return -1

print(linear_search(arr, 32))

#binary search for 45
def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

sorted_arr = sorted(arr)
print(binary_search(sorted_arr, 45))

#Bubble Sort in ascending order

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

print(bubble_sort(arr))

```

```python
#A
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

#B
class LinkedList:
    def __init__(self):
        self.head = None

    # i. Insert at beginning
    def insert_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    # ii. Insert at end
    def insert_end(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return

        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node

    # iii. Display elements
    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")

ll = LinkedList()
ll.insert_beginning(15) # Corrected method name
ll.insert_end(20)
ll.insert_end(40)
ll.display()
```

```python
# question 5
# The data structure is QUEUE
# A queue follows the FIFO (First In, First Out) principle. This enables students to be served according to how they have arrived.

class StudentQueue:
    def __init__(self):
        self.queue = []

    # i. Add student to queue
    def add_student(self, name):
        self.queue.append(name)
        print(name, "added to the queue.")

    # ii. Serve first student
    def serve_student(self):
        if self.queue:
            served = self.queue.pop(0)
            print(served, "has been served.")
        else:
            print("Queue is empty.")

    # iii. Display queue
    def display_queue(self):
        if self.queue:
            print("Current queue:", self.queue)
        else:
            print("Queue is empty.")


# simulation
finance = StudentQueue() # Corrected class name
finance.add_student("Allein")
finance.add_student("Palmer")
finance.add_student("brayo")

finance.display_queue()
finance.serve_student()
finance.display_queue()
```
